# Подключение библиотек NumPy

```python
import numpy as np
```

# Методы создания массивов

```python
arr1 = np.array([10, 20, 30, 40, 50])
print("Из списка:", arr1)

zeros = np.zeros(7)
print("Массив нулей:", zeros)

ones_2d = np.ones((3, 5))
print("Массив единиц 3×5:\n", ones_2d)

seq = np.arange(0, 30, 5)
print("arange(0, 30, 5):", seq)
```

# Атрибуты массивов

```python
print("Форма ones_2d:", ones_2d.shape)
print("Тип данных ones_2d:", ones_2d.dtype)
print("Форма seq:", seq.shape)
print("Тип данных seq:", seq.dtype)
```

# Работа с индексами и срезами

```python
print("Первый элемент arr1:", arr1[0])
print("Последний элемент arr1:", arr1[-1])

print("Элементы arr1 со 2-го по 4-й:", arr1[1:4])

print("Элемент [1, 3] в ones_2d:", ones_2d[1, 3])

print("Подмассив 2×3 из ones_2d:\n", ones_2d[:2, 1:4])
```

# Векторные операции и статистика

```python
a = np.array([1, 2, 3, 4, 5])
b = np.array([10, 20, 30, 40, 50])

print("a + b :", a + b)
print("a * b :", a * b)
print("a в квадрате :", a ** 2)

print("Среднее значение a:", np.mean(a))
print("Сумма элементов a:", np.sum(a))
print("Минимум и максимум a:", np.min(a), "и", np.max(a))
```
